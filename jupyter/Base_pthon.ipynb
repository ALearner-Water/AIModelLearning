```python
#运算符
a=10
b=3
print(a/b)  #除法，结果为浮点数
print(a//b)  #整除，结果为整数
print(a%b)  #取余，结果为整数
print(a**b)  #幂运算，结果为整数
```

```python
#字符串拓展
name1 = "张三"
name2= '张三'
name3 = """张三"""

#字符串包含引号也可以使用转移自符\
str1="'你好'"
str2='"你好"'
str3="\"你好"
print(str3)

#字符串拼接
print(name1+name2+name3)  #字符串拼接 只能拼字符串

#字符串格式化 （占位符拼接）
age=18
year=2024
page=103.6
Name="张三"
print("我的名字是%s" % Name) #字符串用%s
print("今年我%d岁,今年是%d年" % (age,year))  #占位符拼接
print("工资为%f元" % page)  #占位符拼接
```

```python
#字符串格式化对数字的精确处理  %2f 宽度为2小于数字本身不生效  %2.3f 宽度为2 小数点后3位
print("工资为%5.1f元" % page)  #占位符

#第二种方法使用f"{}"格式化字符串
print(f"我的名字是{Name},今年{age}岁")  #不管类型和精度
```

```python
#对上面的总结练习 定义变量进行计算然后输出
name="中国平安"
stock_price=19.99
stock_num=1000
growth_factor=1.2
day=7
print(f"公司：{name} 股票代码：{stock_num} 股票价格：{stock_price}元")
print("每日增长系数是%.1f,经过%d天后价格变为%.2f元"%(growth_factor,day,growth_factor**day*stock_price))
```

```python
#input()函数获取用户输入
name=input("请输入你的名字\n")
print(f"你的名字是{name}")  #输出用户输入的名字
number=(int(input("请输入一个数字\n")))
print(f"你输入的数字的类型是{type(number)}")  #input()函数获取的输入都是字符串类型 计算需要自己强转
```
