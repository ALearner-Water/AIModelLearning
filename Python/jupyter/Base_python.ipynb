```python
#猜数字
import random
num=random.randint(1,10)  #随机生成1-10的整数
print("有三次机会猜数字")
guess_num=int(input("请输入你猜的数字\n"))
if guess_num!=num:
    if guess_num>num:
        print("你猜的数字大了")
    else:
        print("你猜的数字小了")
    guess_num=int(input("请再猜一次\n"))
    if guess_num!=num:
        if guess_num>num:
            print("你猜的数字大了")
        else:
            print("你猜的数字小了")
        guess_num=int(input("请再猜一次\n"))
        if guess_num!=num:
            if guess_num>num:
                print("你猜的数字大了")
            else:
                print("你猜的数字小了")
            print(f"很遗憾，你没有猜对，正确答案是{num}")
        else:
            print("恭喜你，猜对了")
    else:
        print("恭喜你，猜对了")
else:
    print("恭喜你，猜对了")
```

```python
#函数 def 函数名 (传参)：
def sum (i,j):
    return i+j
print(sum(6,7)) #需要写在函数定义之后

#global关键字修改全局变量
a=10
def num():
    global a
    a=100
print(f"修改之前a的值为{a}")
num()   #调用函数之后全局变量被改变
print(f"修改之后a的值为{a}")
```

```python
#函数练习(ATM取款)
flag=True
money=5000000
name=input("请输入姓名:")
def search (name,money):
    print(f"{name}你好,余额剩余{money}元")

def storage (name):
    stroage_money=int(input("请输入金额："))
    print(f"{name},你好,已存入{stroage_money}元")
    global money
    money+=stroage_money
    search(name,money)

def get (name):
    get_money=int(input("请输入金额："))
    print(f"{name},你好,已取出{get_money}元")
    global money
    money=money-get_money
    search(name,money)

def myexit ():
    global flag
    flag=False

while(flag):
    choose=int(input(f"{name},这里是ATM\n,1是查询余额\n,2是存款\n,3是取款\n,4是退出\n"))
    if choose==1:
        search (name,money)
    elif choose==2:
        storage (name)
    elif choose==3:
        get (name)
    elif choose==4 :
        myexit()
print("你已退出")
```
