```python
#函数练习(ATM取款)
flag=True
money=5000000
name=input("请输入姓名:")
def search (name,money):
    print(f"{name}你好,余额剩余{money}元")

def storage (name):
    stroage_money=int(input("请输入金额："))
    print(f"{name},你好,已存入{stroage_money}元")
    global money
    money+=stroage_money
    search(name,money)

def get (name):
    get_money=int(input("请输入金额："))
    print(f"{name},你好,已取出{get_money}元")
    global money
    money=money-get_money
    search(name,money)

def myexit ():
    global flag
    flag=False

while(flag):
    choose=int(input(f"{name},这里是ATM\n,1是查询余额\n,2是存款\n,3是取款\n,4是退出\n"))
    if choose==1:
        search (name,money)
    elif choose==2:
        storage (name)
    elif choose==3:
        get (name)
    elif choose==4 :
        myexit()
print("你已退出")
```

```python
#集合 set 不支持重复元素且无序,无法使用下标访问
set1={"nihao","nihao","jkl","jkl","hello"} #会自动去重  
set2=set()  #创建空集合
print(set1)
print(set2)

#添加元素 add
set1.add("jkl") #无法添加 
set1.add("qwe")
print(set1)

#移除元素 remove
set1.remove("jkl")
print(set1)

#随机取出一个元素 pop
print(set1.pop())
print(set1)     #取出后相当于移除这个元素

#清空集合
set1.clear()
print(set1)

#取出两个元素的差集 生成新集合原集合不变
set3={12,1,2,3,4,5}
set4={12,4,5,6,7,8}
set5=set3.difference(set4)  #集合3有集合4没有
print(set5)

#留下差集
set3.difference_update(set4)    #集合3被修改，集合4不变
print(set3)

#合并集合 union 生成新集合
set6=set3.union(set4)
print(set6)

#可以用for遍历while不可以，有len()可知道集合数量
for x in set6:
    print(x,end=" ")
```

```python
# 函数的多种传参方式
# 位置参数 普通的函数
# 关键字参数  function(x=2,y=3) 直接键=值    可以不按顺序 function(y=3,x=2)
# 缺省参数，默认参数 def function(x,y=3)
print(function(1))  # 此时y可以不传，不传则按默认参数继续使用 设置默认值必须统一放在最后
# 不定长参数/可变参数
def fun1(*args,):    # 位置传递   def fun(*args):    可以传任意个参数，被args收集成元组 tuple
    print(args)
dict={1:"h",2:"y"}
fun1(dict)   #把列表当成元素包裹在元组里面
fun1(1,2,3,4,5,8,9)
fun1()   #传参可以为空
def fun2(**args):   #关键字传递 def fun(**args)     使用key=value的形式传，被args收集成字典 dict
    print(args)
fun2(name="xioamin",age=18,gender="boy")
fun2() 
```

```python
# 匿名函数
# 1.将函数作为参数传递
def coun(funs):     #这是运算逻辑的传递，不是值的传递
    return funs(1,2)    #funs的运算逻辑可以随时改变，但是函数的嵌套就固定下来了
def myadd(x,y):     #可以指定任意运算
    return x+y
def mylose(x,y):
    return x-y
print(coun(myadd))
print(coun(mylose))
# 2.lambda表达式 定义匿名名称，只使用一次 lambda 参数：代码,只可以写一行
lambda x, y: x + y
print(coun(lambda x, y: x**y))  # 只能使用一次
```

```python
# 对文件的操作 编码 读取 写入 追加
# 编码 就是将内容通过密码本(UTF-8)来编写成二进制
# 文件的读取 打开文件 open(文件位置,操作方式,encoding=编码方式)  "r"是只读 "w"是写入 "a"是追加 若文件不存在则会创建新文件
f=open("D:/53507/PythonAIModelLearning/test.txt","r",encoding="UTF-8")
print(type(f))  # <class '_io.TextIOWrapper'>

#读取后可以执行一系列操作 f.read(字节) 读取多少个字节 ()表示读取全部内容 f.readline() 读取一行 f.readlines() 读取多行并存在列表里面
#若多次读取，是从上一次读取的结尾接着读
read1=f.read(2)
print(read1,type(read1))
read2=f.readline()      #返回的都是字符串
print(read2,type(read2))
read3=f.readlines()     #会把读到的东西存在列表中
print(read3,type(read3))
#使用for循环也可以读
for x in f:
    print(x)

#关闭文件 如果没有关闭，则程序会一直被占用
f.close()

#可以使用with open() as f：的方式来让文件处理之后自动关闭
```
